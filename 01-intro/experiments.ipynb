FULL CODE in a single cell 

```python
asian_cars = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X = asian_cars.values
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w.sum()
```
